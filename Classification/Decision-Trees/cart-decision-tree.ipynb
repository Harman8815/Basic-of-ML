# CART Decision Tree

CART (Classification and Regression Trees) is a decision tree algorithm that can handle both classification and regression tasks. It uses the Gini impurity for classification and mean squared error for regression.

## Key Concepts:

- **Gini Impurity**: Uses Gini index for splitting in classification tasks
- **Binary Splits**: Creates binary splits (only two branches per node)
- **Both Tasks**: Handles both classification and regression
- **Cost-Complexity Pruning**: Uses cost-complexity pruning to prevent overfitting
- **Feature Selection**: Can handle both categorical and continuous features

## When to Use:

- When you need both classification and regression capabilities
- When you prefer binary splits over multi-way splits
- When you need cost-complexity pruning
- When working with mixed data types

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris, make_classification, make_regression
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor, plot_tree
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.preprocessing import StandardScaler

# Set style for better visualizations
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
```

## Part 1: CART for Classification

First, let's explore CART for classification tasks using the Gini impurity criterion.

```python
# Load Iris dataset
iris = load_iris()
X_class = iris.data
y_class = iris.target
feature_names = iris.feature_names
target_names = iris.target_names

print(f"Feature shape: {X_class.shape}")
print(f"Classes: {target_names}")
print(f"Features: {feature_names}")
print(f"\nClass distribution: {np.bincount(y_class)}")
```

```python
# Split the data
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_class, y_class, test_size=0.3, random_state=42, stratify=y_class
)

print(f"Training samples: {X_train_c.shape[0]}")
print(f"Test samples: {X_test_c.shape[0]}")
```

## Train CART Classification Tree (Gini Impurity)

```python
# Initialize CART classifier with Gini impurity
cart_class = DecisionTreeClassifier(
    criterion='gini',  # CART uses Gini impurity
    splitter='best',
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42
)

# Train the model
cart_class.fit(X_train_c, y_train_c)

# Make predictions
y_pred_c = cart_class.predict(X_test_c)

# Calculate accuracy
accuracy_c = accuracy_score(y_test_c, y_pred_c)
print(f"Accuracy: {accuracy_c:.4f}")
print(f"Tree depth: {cart_class.get_depth()}")
print(f"Number of leaves: {cart_class.get_n_leaves()}")
```

## Visualize CART Classification Tree

```python
# Plot the decision tree
plt.figure(figsize=(20, 10))
plot_tree(cart_class, 
          feature_names=feature_names,
          class_names=target_names,
          filled=True,
          rounded=True,
          fontsize=10)
plt.title('CART Classification Tree (Gini Impurity) - Iris Dataset', fontsize=16)
plt.tight_layout()
plt.show()
```

## Classification Evaluation

```python
# Classification report
print("Classification Report:")
print(classification_report(y_test_c, y_pred_c, target_names=target_names))
```

```python
# Confusion Matrix
cm_c = confusion_matrix(y_test_c, y_pred_c)

plt.figure(figsize=(8, 6))
sns.heatmap(cm_c, annot=True, fmt='d', cmap='Blues',
            xticklabels=target_names,
            yticklabels=target_names)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix - CART Classification')
plt.tight_layout()
plt.show()
```

## Feature Importance (Classification)

```python
# Get feature importance
feature_importance_c = cart_class.feature_importances_

# Create DataFrame for visualization
importance_df_c = pd.DataFrame({
    'Feature': feature_names,
    'Importance': feature_importance_c
}).sort_values('Importance', ascending=False)

print("Feature Importance (Classification):")
print(importance_df_c)

# Plot feature importance
plt.figure(figsize=(10, 6))
sns.barplot(data=importance_df_c, x='Importance', y='Feature')
plt.title('Feature Importance - CART Classification')
plt.xlabel('Importance')
plt.tight_layout()
plt.show()
```

## Part 2: CART for Regression

Now let's explore CART for regression tasks using mean squared error.

```python
# Create synthetic regression dataset
X_reg, y_reg = make_regression(
    n_samples=500,
    n_features=5,
    n_informative=3,
    noise=10,
    random_state=42
)

print(f"Regression dataset shape: {X_reg.shape}")
print(f"Target range: [{y_reg.min():.2f}, {y_reg.max():.2f}]")
```

```python
# Split the data
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_reg, y_reg, test_size=0.3, random_state=42
)

print(f"Training samples: {X_train_r.shape[0]}")
print(f"Test samples: {X_test_r.shape[0]}")
```

## Train CART Regression Tree (MSE)

```python
# Initialize CART regressor with MSE
cart_reg = DecisionTreeRegressor(
    criterion='squared_error',  # CART uses MSE for regression
    splitter='best',
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42
)

# Train the model
cart_reg.fit(X_train_r, y_train_r)

# Make predictions
y_pred_r = cart_reg.predict(X_test_r)

# Calculate metrics
mse = mean_squared_error(y_test_r, y_pred_r)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test_r, y_pred_r)
r2 = r2_score(y_test_r, y_pred_r)

print(f"Mean Squared Error: {mse:.4f}")
print(f"Root Mean Squared Error: {rmse:.4f}")
print(f"Mean Absolute Error: {mae:.4f}")
print(f"R² Score: {r2:.4f}")
print(f"Tree depth: {cart_reg.get_depth()}")
print(f"Number of leaves: {cart_reg.get_n_leaves()}")
```

## Visualize CART Regression Tree

```python
# Plot the regression tree
plt.figure(figsize=(20, 10))
plot_tree(cart_reg, 
          feature_names=[f'Feature_{i}' for i in range(X_reg.shape[1])],
          filled=True,
          rounded=True,
          fontsize=8,
          max_depth=3)  # Show only first 3 levels for clarity
plt.title('CART Regression Tree (MSE) - First 3 Levels', fontsize=16)
plt.tight_layout()
plt.show()
```

## Regression Evaluation: Actual vs Predicted

```python
# Plot actual vs predicted
plt.figure(figsize=(10, 6))
plt.scatter(y_test_r, y_pred_r, alpha=0.6)
plt.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'r--', lw=2)
plt.xlabel('Actual Values')
plt.ylabel('Predicted Values')
plt.title('Actual vs Predicted - CART Regression')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Plot residuals
residuals = y_test_r - y_pred_r

plt.figure(figsize=(10, 6))
plt.scatter(y_pred_r, residuals, alpha=0.6)
plt.axhline(y=0, color='r', linestyle='--', lw=2)
plt.xlabel('Predicted Values')
plt.ylabel('Residuals')
plt.title('Residual Plot - CART Regression')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Feature Importance (Regression)

```python
# Get feature importance for regression
feature_importance_r = cart_reg.feature_importances_

# Create DataFrame for visualization
importance_df_r = pd.DataFrame({
    'Feature': [f'Feature_{i}' for i in range(X_reg.shape[1])],
    'Importance': feature_importance_r
}).sort_values('Importance', ascending=False)

print("Feature Importance (Regression):")
print(importance_df_r)

# Plot feature importance
plt.figure(figsize=(10, 6))
sns.barplot(data=importance_df_r, x='Importance', y='Feature')
plt.title('Feature Importance - CART Regression')
plt.xlabel('Importance')
plt.tight_layout()
plt.show()
```

## Cost-Complexity Pruning

CART uses cost-complexity pruning to prevent overfitting. Let's explore this.

```python
# Get cost-complexity pruning path
path = cart_class.cost_complexity_pruning_path(X_train_c, y_train_c)
ccp_alphas, impurities = path.ccp_alphas, path.impurities

print(f"Number of different alpha values: {len(ccp_alphas)}")
print(f"Alpha range: [{ccp_alphas.min():.6f}, {ccp_alphas.max():.6f}]")
```

```python
# Plot impurity vs alpha
plt.figure(figsize=(10, 6))
plt.plot(ccp_alphas[:-1], impurities[:-1], marker='o', drawstyle="steps-post")
plt.xlabel('Effective Alpha')
plt.ylabel('Total Impurity of Leaves')
plt.title('Total Impurity vs Effective Alpha for Training Set')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Train trees for different alphas
clfs = []
for ccp_alpha in ccp_alphas:
    clf = DecisionTreeClassifier(criterion='gini', random_state=42, ccp_alpha=ccp_alpha)
    clf.fit(X_train_c, y_train_c)
    clfs.append(clf)

# Remove the last tree (only root node)
clfs = clfs[:-1]
ccp_alphas = ccp_alphas[:-1]

print(f"Number of trees after removing last: {len(clfs)}")
```

```python
# Calculate scores for different alphas
train_scores = [clf.score(X_train_c, y_train_c) for clf in clfs]
test_scores = [clf.score(X_test_c, y_test_c) for clf in clfs]

node_counts = [clf.tree_.node_count for clf in clfs]
depth = [clf.tree_.max_depth for clf in clfs]

fig, ax = plt.subplots(2, 1, figsize=(10, 12))

# Plot accuracy vs alpha
ax[0].plot(ccp_alphas, train_scores, marker='o', label='Train', drawstyle="steps-post")
ax[0].plot(ccp_alphas, test_scores, marker='o', label='Test', drawstyle="steps-post")
ax[0].set_xlabel('Alpha')
ax[0].set_ylabel('Accuracy')
ax[0].set_title('Accuracy vs Alpha')
ax[0].legend()
ax[0].grid(True, alpha=0.3)

# Plot node count vs alpha
ax[1].plot(ccp_alphas, node_counts, marker='o', drawstyle="steps-post")
ax[1].set_xlabel('Alpha')
ax[1].set_ylabel('Number of Nodes')
ax[1].set_title('Number of Nodes vs Alpha')
ax[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Find best alpha
best_alpha_idx = np.argmax(test_scores)
best_alpha = ccp_alphas[best_alpha_idx]
best_clf = clfs[best_alpha_idx]

print(f"Best alpha: {best_alpha:.6f}")
print(f"Best test accuracy: {test_scores[best_alpha_idx]:.4f}")
print(f"Tree depth with best alpha: {best_clf.get_depth()}")
print(f"Number of leaves with best alpha: {best_clf.get_n_leaves()}")
```

## Cross-Validation

```python
# Cross-validation for classification
cv_scores_class = cross_val_score(cart_class, X_train_c, y_train_c, cv=5)
print(f"Classification CV scores: {cv_scores_class}")
print(f"Mean CV accuracy: {cv_scores_class.mean():.4f} (+/- {cv_scores_class.std() * 2:.4f})")

# Cross-validation for regression
cv_scores_reg = cross_val_score(cart_reg, X_train_r, y_train_r, cv=5, scoring='r2')
print(f"\nRegression CV scores: {cv_scores_reg}")
print(f"Mean CV R²: {cv_scores_reg.mean():.4f} (+/- {cv_scores_reg.std() * 2:.4f})")
```

## Comparison: Gini vs Entropy (Classification)

```python
# Compare Gini (CART) vs Entropy (ID3/C4.5)
cart_gini = DecisionTreeClassifier(criterion='gini', random_state=42)
cart_entropy = DecisionTreeClassifier(criterion='entropy', random_state=42)

cart_gini.fit(X_train_c, y_train_c)
cart_entropy.fit(X_train_c, y_train_c)

y_pred_gini = cart_gini.predict(X_test_c)
y_pred_entropy = cart_entropy.predict(X_test_c)

acc_gini = accuracy_score(y_test_c, y_pred_gini)
acc_entropy = accuracy_score(y_test_c, y_pred_entropy)

print(f"Gini (CART) Accuracy: {acc_gini:.4f}")
print(f"Entropy (ID3/C4.5) Accuracy: {acc_entropy:.4f}")
print(f"\nGini tree depth: {cart_gini.get_depth()}")
print(f"Entropy tree depth: {cart_entropy.get_depth()}")
```

## Predict on New Data

```python
# Classification prediction
new_sample_class = [[5.1, 3.5, 1.4, 0.2]]
pred_class = cart_class.predict(new_sample_class)[0]
proba_class = cart_class.predict_proba(new_sample_class)[0]

print("Classification Prediction:")
print(f"Sample: {new_sample_class[0]}")
print(f"Predicted class: {target_names[pred_class]}")
print("Probabilities:")
for name, prob in zip(target_names, proba_class):
    print(f"  {name}: {prob:.4f}")

# Regression prediction
new_sample_reg = [[0.5, 0.3, 0.8, 0.1, 0.6]]
pred_reg = cart_reg.predict(new_sample_reg)[0]

print(f"\nRegression Prediction:")
print(f"Sample: {new_sample_reg[0]}")
print(f"Predicted value: {pred_reg:.4f}")
```

## Summary

### Key Takeaways:

1. **Gini Impurity**: CART uses Gini index for classification splitting
2. **Binary Splits**: Creates only binary splits (two branches per node)
3. **Dual Purpose**: Handles both classification and regression tasks
4. **Cost-Complexity Pruning**: Uses alpha parameter for pruning
5. **MSE for Regression**: Uses mean squared error for regression splitting

### Advantages:
- Handles both classification and regression
- Binary splits are simpler and more interpretable
- Cost-complexity pruning prevents overfitting
- Handles both categorical and continuous features
- No need for feature scaling
- Provides feature importance

### Limitations:
- Can overfit without proper pruning
- Binary splits may require deeper trees
- Decision boundaries are axis-aligned
- Can be unstable (small data changes affect tree)
- May not perform well on linearly separable data

### CART vs ID3/C4.5:

- **CART**: Binary splits, Gini impurity, handles regression, cost-complexity pruning
- **ID3**: Multi-way splits, information gain, categorical only, no pruning
- **C4.5**: Multi-way splits, gain ratio, handles continuous, includes pruning
